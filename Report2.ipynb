```python
# 5x5 크기의 행렬을 사용자로부터 받는 함수
def input_relation_matrix():
    matrix = []
    print("5x5 관계 행렬을 한 줄씩 입력하세요 (각 줄에 0 또는 1이 5개, 공백 구분):")
    for i in range(5):
        while True:
            row = input(f"{i+1}번째 행: ").split()
            if len(row) == 5 and all(x in ['0','1'] for x in row):
                matrix.append([int(x) for x in row])
                break
            else:
                print("다시 0과 1로 이루어진 5개의 수를 입력하세요.")
    return matrix
```

```python
# 1. 관계 행렬 입력 (공통 함수 사용)
mat = input_relation_matrix()

# 2. 성질 판별 함수
def is_reflexive(m):
    return all(m[i][i] == 1 for i in range(5))
def is_symmetric(m):
    return all(m[i][j] == m[j][i] for i in range(5) for j in range(5))
def is_transitive(m):
    return all(m[i][j] != 1 or m[j][k] != 1 or m[i][k] == 1 for i in range(5) for j in range(5) for k in range(5))

# 3. 판별 및 결과 출력
ref = is_reflexive(mat)
sym = is_symmetric(mat)
tra = is_transitive(mat)
print(f"반사성: {'예' if ref else '아니오'}")
print(f"대칭성: {'예' if sym else '아니오'}")
print(f"추이성: {'예' if tra else '아니오'}")
if ref and sym and tra:
    print("이 관계는 동치 관계입니다.")
else:
    print("이 관계는 동치 관계가 아닙니다.")
```

```python
mat = input_relation_matrix()

# 폐포 함수들
def reflexive_closure(m):
    res = [r[:] for r in m]
    for i in range(5):
        res[i][i] = 1
    return res
def symmetric_closure(m):
    res = [r[:] for r in m]
    for i in range(5):
        for j in range(5):
            if res[i][j]:
                res[j][i] = 1
    return res
def transitive_closure(m):
    res = [r[:] for r in m]
    for k in range(5):
        for i in range(5):
            for j in range(5):
                if res[i][k] and res[k][j]:
                    res[i][j] = 1
    return res

# 변환·출력 함수
def print_matrix(mat, title):
    print(title)
    for row in mat:
        print(row)

def is_reflexive(m):
    return all(m[i][i] == 1 for i in range(5))
def is_symmetric(m):
    return all(m[i][j] == m[j][i] for i in range(5) for j in range(5))
def is_transitive(m):
    return all(m[i][j] != 1 or m[j][k] != 1 or m[i][k] == 1 for i in range(5) for j in range(5) for k in range(5))

def print_equiv_classes(m):
    for i in range(5):
        eq = [j+1 for j in range(5) if m[i][j]==1]
        print(f"원소 {i+1}의 동치류: {{ {' ,'.join(map(str, eq))} }}")

# 단계별 실행
print_matrix(mat, "변환 전 행렬:")
if not is_reflexive(mat):
    mat = reflexive_closure(mat)
    print_matrix(mat, "반사 폐포 적용 후:")
if not is_symmetric(mat):
    mat = symmetric_closure(mat)
    print_matrix(mat, "대칭 폐포 적용 후:")
if not is_transitive(mat):
    mat = transitive_closure(mat)
    print_matrix(mat, "추이 폐포 적용 후:")

# 최종 동치 관계, 동치류 판정
if is_reflexive(mat) and is_symmetric(mat) and is_transitive(mat):
    print("최종 행렬은 동치 관계입니다.\n동치류:")
    print_equiv_classes(mat)
else:
    print("최종 행렬도 동치 관계가 아닙니다.")
```
